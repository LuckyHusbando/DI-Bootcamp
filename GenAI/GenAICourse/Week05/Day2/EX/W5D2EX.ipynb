```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import pandas as pd
from scipy.stats import norm, ttest_ind, linregress, f_oneway, binom
import scipy

# ሴ Exercise 1: Basic Usage of SciPy
print(f"Exercise 1: SciPy Version: {scipy.__version__}")

# ሴ Exercise 2: Descriptive Statistics
data_ex2 = [12, 15, 13, 12, 18, 20, 22, 21]
print(f"\nExercise 2:")
print(f"Mean: {np.mean(data_ex2)}")
print(f"Median: {np.median(data_ex2)}")
print(f"Variance: {np.var(data_ex2, ddof=1):.2f}")
print(f"Std Deviation: {np.std(data_ex2, ddof=1):.2f}")

# ሴ Exercise 3: Understanding Distributions
print("\nExercise 3: Generating plot...")
mu, sigma = 50, 10
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)
plt.plot(x, norm.pdf(x, mu, sigma))
plt.title("Normal Distribution (Mean=50, Std=10)")
plt.show()

# ሴ Exercise 4: T-Test Application
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)
t_stat, p_val = ttest_ind(data1, data2)
print(f"\nExercise 4: T-Test Results")
print(f"T-statistic: {t_stat:.4f}, P-value: {p_val:.4f}")

# ሴ Exercise 5: Linear Regression Analysis
house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]
slope, intercept, r_value, p_value, std_err = linregress(house_sizes, house_prices)
predicted_90 = slope * 90 + intercept
print(f"\nExercise 5: Linear Regression")
print(f"Slope: {slope:.2f}, Intercept: {intercept:.2f}")
print(f"Predicted price for 90 sqm: {predicted_90:.2f}")
print(f"Interpretation: For every 1 sqm increase in size, the price increases by {slope:.2f} units.")

# ሴ Exercise 6: Understanding ANOVA
fert1 = [5, 6, 7, 6, 5]
fert2 = [7, 8, 7, 9, 8]
fert3 = [4, 5, 4, 3, 4]
f_stat, p_anova = f_oneway(fert1, fert2, fert3)
print(f"\nExercise 6: ANOVA Results")
print(f"F-value: {f_stat:.4f}, P-value: {p_anova:.4f}")
if p_anova < 0.05:
    print("Result: The fertilizers have significantly different effects (p < 0.05).")
else:
    print("Result: No significant difference found (p > 0.05).")

# ሴ Exercise 7: Probability Distributions (Optional)
prob_5_heads = binom.pmf(5, 10, 0.5)
print(f"\nExercise 7: Binomial Distribution")
print(f"Probability of exactly 5 heads in 10 flips: {prob_5_heads:.4f}")

# ሴ Exercise 8: Correlation Coefficients (Optional)
df = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})
pearson_corr, _ = stats.pearsonr(df['age'], df['income'])
spearman_corr, _ = stats.spearmanr(df['age'], df['income'])
print(f"\nExercise 8: Correlation")
print(f"Pearson Correlation: {pearson_corr:.4f}")
print(f"Spearman Correlation: {spearman_corr:.4f}")
```
