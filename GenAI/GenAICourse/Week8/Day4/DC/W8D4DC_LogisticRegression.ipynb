#Daily Challenge: Logistic Regression for Admission Prediction

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Data Exploration
# Assuming ex2data1.txt is comma-separated: Exam 1, Exam 2, Admitted
data = pd.read_csv('/content/ex2data1.txt', header=None, names=['Exam 1', 'Exam 2', 'Admitted'])
print("First few rows of the dataset:")
print(data.head())

# Visualization
plt.figure(figsize=(10, 6))
admitted = data[data['Admitted'] == 1]
not_admitted = data[data['Admitted'] == 0]

plt.scatter(admitted['Exam 1'], admitted['Exam 2'], marker='+', c='b', label='Admitted')
plt.scatter(not_admitted['Exam 1'], not_admitted['Exam 2'], marker='o', c='r', label='Not Admitted')
plt.xlabel('Exam 1 Score')
plt.ylabel('Exam 2 Score')
plt.legend()
plt.title('Scatter plot of training data')
plt.show()

# 2. Applying Logistic Regression with scikit-learn
X = data[['Exam 1', 'Exam 2']]
y = data['Admitted']

model = LogisticRegression()
model.fit(X, y)

# 3. Making Predictions & Evaluation
y_pred = model.predict(X)
accuracy = accuracy_score(y, y_pred)

print(f"Model Parameters: Intercept: {model.intercept_[0]}, Coefficients: {model.coef_[0]}")
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
