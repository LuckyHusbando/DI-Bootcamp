```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. Data Import with specific encoding to fix UnicodeDecodeError
file_path = '/content/Airplane_Crashes_and_Fatalities_Since_1908_t0_2023.csv'
df = pd.read_csv(file_path, encoding='latin1')

# 2. Data Cleaning
# Convert Date to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Extract Year for trend analysis
df['Year'] = df['Date'].dt.year

# Handling missing values: Fill numeric with 0, categorical with 'Unknown'
df['Fatalities'] = df['Fatalities'].fillna(0)
df['Aboard'] = df['Aboard'].fillna(0)
df['Operator'] = df['Operator'].fillna('Unknown')

# Calculate Survival Rate
df['Survival_Rate'] = 1 - (df['Fatalities'] / df['Aboard'])
df['Survival_Rate'] = df['Survival_Rate'].replace([np.inf, -np.inf], 0).fillna(0)

print("Dataset imported and cleaned successfully.")
print(f"Total records: {len(df)}")
print(df.head())
```
