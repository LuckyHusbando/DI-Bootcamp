# Part 4: Custom Attention Implementation
Build the simple attention layer, classifier, and data pipeline for the scratch model.


> **Learning point**
> Scaling the dot products by $1/\sqrt{d_k}$ keeps gradients stable and prevents the softmax from collapsing when embeddings grow. This opeeration is crucial for training deep attention models.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset


class Attention(nn.Module):
    def __init__(self, embed_dim):
        super().__init__()
        self.scale = embed_dim**-0.5  # Scaled dot-product scaling factor

    def forward(self, query, key, value, mask=None):
        # Transpose last two dimensions of key: (batch, seq_len, embed_dim) -> (batch, embed_dim, seq_len)
        scores = torch.matmul(query, key.transpose(-2, -1)) * self.scale

        if mask is not None:
            scores = scores.masked_fill(mask == 0, float("-inf"))

        # Apply softmax across the key sequence dimension (last dim)
        attn = F.softmax(scores, dim=-1)

        # Compute weighted sum of values
        return torch.matmul(attn, value), attn


class SimpleAttentionClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.attn = Attention(embed_dim)
        self.fc = nn.Linear(embed_dim, num_classes)

    def forward(self, x):
        embed = self.embedding(x)

        # Self-attention: query = key = value = embed
        attn_output, _ = self.attn(embed, embed, embed)

        # Mean pooling across the sequence dimension (dim=1)
        pooled = attn_output.mean(dim=1)

        return self.fc(pooled)
```

```python
# TODO: preprocess datasets for the custom attention model
ATTN_TEXT_COLUMN = 'sms'  # TODO: set to 'sms'
ATTN_MAX_LEN = 64      # TODO: set to 64
if ATTN_TEXT_COLUMN is None or ATTN_MAX_LEN is None:
    raise ValueError('Complete ATTN_TEXT_COLUMN and ATTN_MAX_LEN.')


def preprocess_for_attention(example):
    tokens = tokenizer.encode(
        example[ATTN_TEXT_COLUMN],
        max_length=ATTN_MAX_LEN,
        truncation=True,
        padding='max_length',
    )
    return {'input_ids': tokens, 'label': example['label']}


train_ds_attn = train_ds.map(preprocess_for_attention)
val_ds_attn = val_ds.map(preprocess_for_attention)
```

### 1. What are the roles of query, key, and value in the attention mechanism?
Query ($Q$): Represents the target token or current position seeking context. It asks, "What information am I looking for?"

Key ($K$): Represents every token's label or visual/semantic tag in the sequence. It answers, "What information do I contain?"

The dot product between a Query and a Key ($Q \cdot K^T$) measures how relevant a specific token is to the current Query.

Value ($V$): Represents the actual content or embedding of each token. Once attention weights are calculated via Query-Key matching, they scale these Values to compute a weighted sum as the final context representation.

Numerical Stability Rationale

Large Magnitude Dot Products: As the key dimension $d_k$ grows large, the standard deviation of the dot products scales by $\sqrt{d_k}$. This causes the magnitude of the raw attention scores to become very large.

Softmax Saturation & Vanishing Gradients: Feeding large values into the softmax function pushes the outputs to extreme probabilities ($1$ for the largest value, $0$ for all others). In these regions, the derivative of softmax $\sigma'(x) = \sigma(x)(1 - \sigma(x))$ approaches zero, causing gradients to vanish and stalling gradient descent during backpropagation.

### 3. How does self-attention differ from traditional sequence models like RNNs?
1. Processing Styl RNNs (Sequential / Recurrent):RNNs process sequences step-by-step ($t_1, t_2, \dots, t_n$). To compute the output at step $t_t$, the network must wait for the hidden state $h_{t-1}$ from the previous step:$$h_t = f(h_{t-1}, x_t)$$Because operations at step $t$ strictly depend on $t-1$, computation cannot be parallelized across the time dimension.

Self-Attention (Parallel / Non-recurrent):Self-attention computes connections between all pairs of tokens simultaneously in a single matrix operation:$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$There is no temporal loop; all token representations are updated concurrently across the GPU. Position information is provided via positional encodings rather than sequential processing.

2. Dependency Capture & Path Length

RNNs ($\mathcal{O}(n)$ Path Length):Information from token $1$ must step through $n-1$ intermediate hidden states to reach token $n$. Over long distances, early information is frequently lost due to the vanishing/exploding gradient problem or information bottlenecking in the hidden state vector.Self-Attention ($\mathcal{O}(1)$

Path Length):Every token directly compares itself against every other token via Query-Key dot products. The path distance between any two tokens in the sequence—regardless of whether they are adjacent or 1,000 tokens apart—is always $1$. This allows Transformers to easily model long-range relationships without information decay.
