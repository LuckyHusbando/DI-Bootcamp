```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("--- Exercise 1: Matrix Operations ---")
matrix = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
det = np.linalg.det(matrix)
inv = np.linalg.inv(matrix)
print(f"Matrix:\n{matrix}")
print(f"Determinant: {det}")
print(f"Inverse:\n{inv}")

print("\n--- Exercise 2: Statistical Analysis ---")
data = np.random.rand(50)
print(f"Mean: {np.mean(data)}")
print(f"Median: {np.median(data)}")
print(f"Standard Deviation: {np.std(data)}")

print("\n--- Exercise 3: Date Manipulation ---")
dates = np.arange('2023-01', '2023-02', dtype='datetime64[D]')
formatted_dates = np.array([pd.to_datetime(d).strftime('%Y/%m/%d') for d in dates])
print(f"First 5 dates: {formatted_dates[:5]}")

print("\n--- Exercise 4: NumPy and Pandas ---")
df = pd.DataFrame(np.random.randn(10, 4), columns=list('ABCD'))
selection = df[df['A'] > 0]
print(f"Rows where A > 0:\n{selection}")
print(f"Sum of columns:\n{df.sum()}")

print("\n--- Exercise 5: Image Representation ---")
# Images are 2D arrays (grayscale) or 3D (RGB). Each value represents pixel intensity.
image = np.random.randint(0, 256, (5, 5), dtype=np.uint8)
print(f"5x5 Grayscale Image Array:\n{image}")

print("\n--- Exercise 6: Basic Hypothesis Testing ---")
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
mean_diff = np.mean(productivity_after) - np.mean(productivity_before)
print(f"Hypothesis: The training program increases productivity.")
print(f"Observed Mean Increase: {mean_diff:.2f}")

print("\n--- Exercise 7: Complex Array Comparison ---")
arr1 = np.array([10, 20, 30])
arr2 = np.array([15, 15, 35])
comparison = arr1 > arr2
print(f"Array 1: {arr1}, Array 2: {arr2}")
print(f"Is Array 1 > Array 2? {comparison}")

print("\n--- Exercise 8: Time Series Slicing ---")
time_series = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
series_data = pd.Series(np.random.randn(len(time_series)), index=time_series)
print(f"Q1 Slice (Jan-Mar) count: {len(series_data['2023-01':'2023-03'])}")

print("\n--- Exercise 9: Data Conversion ---")
np_arr = np.array([[1, 2], [3, 4]])
pd_df = pd.DataFrame(np_arr)
back_to_np = pd_df.to_numpy()
print(f"Back to NumPy:\n{back_to_np}")

print("\n--- Exercise 10: Basic Visualization ---")
plt.plot(np.random.randn(50).cumsum())
plt.title("Random Walk Example")
plt.show()
```
