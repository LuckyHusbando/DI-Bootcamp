Use NumPy to generate a synthetic dataset representing average monthly temperatures (in degrees Celsius) for 12 months across 10 different cities. The temperatures should range from -5 to 35 degrees.

Convert this NumPy array into a Pandas DataFrame, adding city names as index and months as columns.


2. Data Analysis:

Hint 1: Calculate the annual average temperature using DataFrame.mean(axis).

Identify the city with the highest and lowest average temperature for the year.

Evaluation Criteria:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Data Preparation
np.random.seed(42) # For reproducibility
cities = ['Tokyo', 'New York', 'London', 'Paris', 'Sydney', 'Mumbai', 'Cairo', 'Moscow', 'Rio', 'Berlin']
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Generate synthetic data: 10 cities, 12 months, range -5 to 35
temp_data = np.random.uniform(-5, 35, size=(10, 12))

# Create DataFrame
df_temp = pd.DataFrame(temp_data, index=cities, columns=months)

# 2. Data Analysis
# Calculate annual average temperature for each city
df_temp['Annual_Avg'] = df_temp.mean(axis=1)

# Identify city with highest and lowest average
highest_city = df_temp['Annual_Avg'].idxmax()
lowest_city = df_temp['Annual_Avg'].idxmin()

print(f"City with highest average temperature: {highest_city} ({df_temp.loc[highest_city, 'Annual_Avg']:.2f}°C)")
print(f"City with lowest average temperature: {lowest_city} ({df_temp.loc[lowest_city, 'Annual_Avg']:.2f}°C)")

# 3. Data Visualization
plt.figure(figsize=(12, 6))
for city in cities:
    plt.plot(months, df_temp.loc[city, months], marker='o', label=city)

plt.title('Monthly Temperature Trends per City')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Brief Report
print("\n--- Summary Report ---")
print(f"The analysis shows that {highest_city} had the warmest year on average, while {lowest_city} was the coolest.")
print("The visualization highlights significant seasonal fluctuations across all cities, reflecting the synthetic variety in the dataset.")
```
